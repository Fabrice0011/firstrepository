```python
FILE_PATH  = "C:\\Users\\user\\Desktop\\DATA SET 2\\student analysis.csv"       # <-- path to your dataset (csv). For excel use pd.read_excel
ID_COL     = "student_id"         # identifier column (not used for analysis/modelling)
TARGET_REG = "exam_score"         # numeric target for regression
TARGET_CLF = ["pass_status", "performance_level"]   # categorical targets for classification
# Columns that are derived from the exam score -> they must NEVER be used as
# features (that would be "data leakage" and give fake, perfect results)
LEAKY_COLS = ["exam_score", "performance_grade", "pass_status", "performance_level"]
RANDOM_STATE = 42
SAVE_PLOTS   = True               # save every figure as png in OUTPUT_DIR
SHOW_PLOTS   = True               # set False to run without popping up windows
OUTPUT_DIR   = "analysis_output"
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)


def finish_plot(name):
    """Save (optional) and show (optional) the current figure, then close it."""
    plt.tight_layout()
    if SAVE_PLOTS:
        plt.savefig(os.path.join(OUTPUT_DIR, f"{name}.png"), bbox_inches="tight")
    if SHOW_PLOTS:
        plt.show()
    plt.close()


def header(text):
    """Print a nice section header."""
    print("\n" + "=" * 80 + f"\n{text}\n" + "=" * 80)
```

```python
#header("1. LOAD DATA & FIRST LOOK")

df = pd.read_csv(FILE_PATH)

print("Shape (rows, columns):", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nColumn types & non-null counts:")
df.info()

print("\nMissing values per column (sum):")
print(df.isnull().sum()[df.isnull().sum() > 0] if df.isnull().any().any() else "No missing values")

# Split columns into numeric / categorical groups (used throughout the script)
num_cols = df.select_dtypes(include="number").columns.tolist()
cat_cols = df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()
# Identifier should not be treated as an analysis variable
num_cols = [c for c in num_cols if c != ID_COL]
cat_cols = [c for c in cat_cols if c != ID_COL]

print(f"\n{len(num_cols)} numeric columns, {len(cat_cols)} categorical columns")
print("Categorical columns:", cat_cols)

print("\nSummary statistics (numeric):")
print(df[num_cols].describe().T.round(2))

print("\nSummary statistics (categorical):")
if cat_cols:
    print(df[cat_cols].describe().T)
```

```python
# 2. DATA CLEANING & SANITY CHECKS
# ------------------------------------------------------------------------------
header("2. DATA CLEANING & SANITY CHECKS")

# 2.1 Duplicate rows and duplicate students
print("Duplicate rows         :", df.duplicated().sum())
if ID_COL in df.columns:
    print("Duplicate student IDs  :", df[ID_COL].duplicated().sum())

# 2.2 Impossible / suspicious values
#     (percentages must be 0-100, hours per day must be 0-24, scores must be >= 0 ...)
pct_like = [c for c in num_cols if any(k in c for k in ["percentage", "rate", "score"])
            and df[c].max() <= 150]
hour_like = [c for c in num_cols if "hours" in c and "per_day" in c or c == "sleep_hours"]

for c in pct_like:
    bad = df[(df[c] < 0) | (df[c] > 100)]
    if len(bad):
        print(f"[WARNING] {c}: {len(bad)} rows outside 0-100")
for c in hour_like:
    bad = df[(df[c] < 0) | (df[c] > 24)]
    if len(bad):
        print(f"[WARNING] {c}: {len(bad)} rows outside 0-24 hours")
print("Range checks finished.")

# 2.3 Fix text columns: strip spaces (e.g. 'Urban ' vs 'Urban')
for c in cat_cols:
    df[c] = df[c].astype(str).str.strip()

# 2.4 Remove exact duplicate rows (if any) and fill any missing values
df = df.drop_duplicates().reset_index(drop=True)
for c in num_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].median())          # numeric -> median
for c in cat_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].mode()[0])         # categorical -> mode

# 2.5 Outlier summary using the IQR rule (we only REPORT, we do not delete)
outlier_rows = []
for c in num_cols:
    q1, q3 = df[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((df[c] < low) | (df[c] > high)).sum()
    outlier_rows.append((c, n_out, round(100 * n_out / len(df), 2)))
outlier_df = (pd.DataFrame(outlier_rows, columns=["column", "n_outliers", "percent"])
              .sort_values("n_outliers", ascending=False))
print("\nOutliers (IQR rule) - top 10:")
print(outlier_df.head(10).to_string(index=False))
```

```python
# 3. EXPLORATORY DATA ANALYSIS (EDA)
# ------------------------------------------------------------------------------
header("3. EXPLORATORY DATA ANALYSIS")

# 3.1 Distribution of every numeric column (histograms)
n = len(num_cols)
ncols = 4
nrows = int(np.ceil(n / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(4.5 * ncols, 3.2 * nrows))
for ax, c in zip(np.array(axes).ravel(), num_cols):
    sns.histplot(df[c], bins=30, kde=True, ax=ax)
    ax.set_title(c)
for ax in np.array(axes).ravel()[n:]:
    ax.axis("off")
plt.suptitle("Distributions of numeric variables", y=1.01, fontsize=16)
finish_plot("03_numeric_distributions")

# 3.2 Boxplots to see outliers (standardised so all fit one chart)
plt.figure(figsize=(max(10, n * 0.6), 6))
sns.boxplot(data=(df[num_cols] - df[num_cols].mean()) / df[num_cols].std(), orient="v")
plt.xticks(rotation=90)
plt.title("Boxplots of standardised numeric variables (outlier view)")
finish_plot("03_boxplots_outliers")

# 3.3 Counts for categorical variables
plot_cats = [c for c in cat_cols if df[c].nunique() <= 12]
ncols = 3
nrows = int(np.ceil(len(plot_cats) / ncols)) if plot_cats else 1
fig, axes = plt.subplots(nrows, ncols, figsize=(5.5 * ncols, 4 * nrows))
for ax, c in zip(np.array(axes).ravel(), plot_cats):
    order = df[c].value_counts().index
    sns.countplot(data=df, x=c, order=order, ax=ax)
    ax.set_title(c)
    ax.tick_params(axis="x", rotation=30)
for ax in np.array(axes).ravel()[len(plot_cats):]:
    ax.axis("off")
plt.suptitle("Category counts", y=1.01, fontsize=16)
finish_plot("03_categorical_counts")

# 3.4 Target balance (is the dataset balanced for classification?)
for t in TARGET_CLF:
    if t in df.columns:
        print(f"\nClass balance for '{t}':")
        print(df[t].value_counts(normalize=True).round(3))

# 3.5 Exam-score distribution
if TARGET_REG in df.columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[TARGET_REG], bins=30, kde=True)
    plt.axvline(df[TARGET_REG].mean(), color="red", linestyle="--", label="mean")
    plt.axvline(df[TARGET_REG].median(), color="green", linestyle="--", label="median")
    plt.legend()
    plt.title("Exam score distribution")
    finish_plot("03_exam_score_distribution")
    print("\nSkewness of exam_score:", round(df[TARGET_REG].skew(), 3))
```

```python
# 4. CORRELATION ANALYSIS
# ------------------------------------------------------------------------------
header("4. CORRELATION ANALYSIS")

corr = df[num_cols].corr()

plt.figure(figsize=(max(10, n * 0.55), max(8, n * 0.45)))
sns.heatmap(corr, cmap="coolwarm", center=0, annot=(n <= 20), fmt=".2f", linewidths=0.3)
plt.title("Correlation heatmap (numeric variables)")
finish_plot("04_correlation_heatmap")

if TARGET_REG in corr.columns:
    target_corr = corr[TARGET_REG].drop(TARGET_REG).sort_values(ascending=False)
    print("Correlation of each numeric variable with exam_score:")
    print(target_corr.round(3))

    plt.figure(figsize=(8, max(4, len(target_corr) * 0.3)))
    target_corr.sort_values().plot(kind="barh", color=np.where(target_corr.sort_values() > 0, "seagreen", "indianred"))
    plt.title("Correlation with exam_score")
    finish_plot("04_correlation_with_target")

# Pairs of features that are almost the same (multicollinearity warning)
high_corr = (corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
             .stack().reset_index())
high_corr.columns = ["feature_1", "feature_2", "correlation"]
high_corr = high_corr[high_corr["correlation"].abs() > 0.8].sort_values("correlation", ascending=False)
print("\nHighly correlated pairs (|r| > 0.8):")
print(high_corr.to_string(index=False) if len(high_corr) else "None")
```

```python
# 5. GROUP COMPARISONS + STATISTICAL TESTS
# ------------------------------------------------------------------------------
header("5. GROUP COMPARISONS + STATISTICAL TESTS")
# Question: does exam_score differ between groups (gender, school_type, urban/rural ...)?
#   2 groups  -> Welch t-test        (+ Mann-Whitney as non-parametric check)
#   3+ groups -> one-way ANOVA       (+ Kruskal-Wallis as non-parametric check)
# p-value < 0.05  => statistically significant difference.

group_vars = [c for c in cat_cols if c not in LEAKY_COLS and 2 <= df[c].nunique() <= 10]
results = []

for c in group_vars:
    groups = [g[TARGET_REG].values for _, g in df.groupby(c)]
    if len(groups) == 2:
        stat, p = stats.ttest_ind(groups[0], groups[1], equal_var=False)
        _, p_np = stats.mannwhitneyu(groups[0], groups[1])
        test = "Welch t-test"
    else:
        stat, p = stats.f_oneway(*groups)
        _, p_np = stats.kruskal(*groups)
        test = "ANOVA"
    # Effect size: eta-squared (how much of the score variance the variable explains)
    grand_mean = df[TARGET_REG].mean()
    ss_between = sum(len(g) * (g.mean() - grand_mean) ** 2 for g in groups)
    ss_total = ((df[TARGET_REG] - grand_mean) ** 2).sum()
    results.append((c, test, stat, p, p_np, ss_between / ss_total))

group_results = (pd.DataFrame(results, columns=["variable", "test", "statistic", "p_value",
                                                "p_nonparametric", "eta_squared"])
                 .sort_values("p_value"))
group_results["significant(0.05)"] = group_results["p_value"] < 0.05
print(group_results.round(4).to_string(index=False))

# Mean exam score for each group
for c in group_vars:
    print(f"\nMean {TARGET_REG} by {c}:")
    print(df.groupby(c)[TARGET_REG].agg(["count", "mean", "std"]).round(2).sort_values("mean", ascending=False))

# Boxplots for each grouping variable
ncols = 3
nrows = int(np.ceil(len(group_vars) / ncols)) if group_vars else 1
fig, axes = plt.subplots(nrows, ncols, figsize=(5.5 * ncols, 4.2 * nrows))
for ax, c in zip(np.array(axes).ravel(), group_vars):
    sns.boxplot(data=df, x=c, y=TARGET_REG, ax=ax)
    ax.set_title(f"{TARGET_REG} by {c}")
    ax.tick_params(axis="x", rotation=30)
for ax in np.array(axes).ravel()[len(group_vars):]:
    ax.axis("off")
finish_plot("05_group_boxplots")

# Chi-square test: is each categorical variable associated with pass/fail?
if "pass_status" in df.columns:
    chi_rows = []
    for c in group_vars:
        table = pd.crosstab(df[c], df["pass_status"])
        chi2, p, dof, _ = stats.chi2_contingency(table)
        cramers_v = np.sqrt(chi2 / (table.values.sum() * (min(table.shape) - 1)))
        chi_rows.append((c, chi2, p, cramers_v))
    chi_df = (pd.DataFrame(chi_rows, columns=["variable", "chi2", "p_value", "cramers_v"])
              .sort_values("p_value"))
    print("\nChi-square test vs pass_status (Cramer's V = strength of association):")
    print(chi_df.round(4).to_string(index=False))

# Specific question: does private tuition help?
if "private_tuition" in df.columns:
    print("\nPrivate tuition vs exam score:")
    print(df.groupby("private_tuition")[TARGET_REG].agg(["count", "mean", "std"]).round(2))
```

```python
# 6. STUDY-HABIT & BEHAVIOUR ANALYSIS
# ------------------------------------------------------------------------------
header("6. STUDY-HABIT & BEHAVIOUR ANALYSIS")

habit_candidates = ["study_hours_per_day", "self_study_hours", "online_learning_hours",
                    "attendance_percentage", "assignment_completion_rate",
                    "class_participation", "practice_tests_completed", "sleep_hours",
                    "previous_exam_score", "previous_gpa", "revision_frequency"]
habit_cols = [c for c in habit_candidates if c in num_cols]

# 6.1 Scatter plot + regression line + Pearson/Spearman for each habit
sample = df.sample(min(len(df), 2000), random_state=RANDOM_STATE)   # sample = faster plots
ncols = 3
nrows = int(np.ceil(len(habit_cols) / ncols)) if habit_cols else 1
fig, axes = plt.subplots(nrows, ncols, figsize=(5.5 * ncols, 4.2 * nrows))
for ax, c in zip(np.array(axes).ravel(), habit_cols):
    sns.regplot(data=sample, x=c, y=TARGET_REG, ax=ax,
                scatter_kws={"alpha": 0.3, "s": 12}, line_kws={"color": "red"})
    r, p = stats.pearsonr(df[c], df[TARGET_REG])
    ax.set_title(f"{c}  (r={r:.2f})")
for ax in np.array(axes).ravel()[len(habit_cols):]:
    ax.axis("off")
finish_plot("06_habits_vs_score")

habit_stats = []
for c in habit_cols:
    r, p = stats.pearsonr(df[c], df[TARGET_REG])
    rho, p_s = stats.spearmanr(df[c], df[TARGET_REG])
    habit_stats.append((c, r, p, rho, p_s))
print(pd.DataFrame(habit_stats, columns=["variable", "pearson_r", "p_pearson", "spearman_rho", "p_spearman"])
      .sort_values("pearson_r", ascending=False).round(4).to_string(index=False))

# 6.2 Categorical habit variables vs score (study method, environment, sleep quality ...)
for c in ["study_method", "study_environment", "sleep_quality", "study_consistency",
          "notes_quality", "revision_frequency"]:
    if c in cat_cols:
        plt.figure(figsize=(8, 4.5))
        order = df.groupby(c)[TARGET_REG].mean().sort_values().index
        sns.barplot(data=df, x=c, y=TARGET_REG, order=order, errorbar="sd")
        plt.xticks(rotation=30)
        plt.title(f"Average {TARGET_REG} by {c}")
        finish_plot(f"06_{c}_vs_score")

# 6.3 Binned analysis: how does the score change as study hours increase?
if "study_hours_per_day" in df.columns:
    df["study_hours_bin"] = pd.cut(df["study_hours_per_day"], bins=5)
    print("\nAverage exam score by study-hours bin:")
    print(df.groupby("study_hours_bin")[TARGET_REG].agg(["count", "mean"]).round(2))
    df = df.drop(columns="study_hours_bin")
```

```python
# 7. REGRESSION MODELS  (predict exam_score)
# ------------------------------------------------------------------------------
#head("7. REGRESSION MODELS")

# 7.1 Build feature matrix. We DROP: id + all leaky columns (derived from exam_score)
drop_cols = [c for c in [ID_COL] + LEAKY_COLS if c in df.columns]
X = df.drop(columns=drop_cols)
y = df[TARGET_REG]

num_features = X.select_dtypes(include="number").columns.tolist()
cat_features = X.select_dtypes(exclude="number").columns.tolist()
print(f"Features used: {X.shape[1]}  ({len(num_features)} numeric, {len(cat_features)} categorical)")


def make_preprocessor(scale_numeric=True):
    """Impute + scale numeric columns, impute + one-hot encode categorical columns."""
    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale_numeric:
        num_steps.append(("scaler", StandardScaler()))
    return ColumnTransformer([
        ("num", Pipeline(num_steps), num_features),
        ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                          ("onehot", OneHotEncoder(handle_unknown="ignore"))]), cat_features),
    ])


# 7.2 Train / test split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)

# 7.3 Models to compare
reg_models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(n_estimators=200, n_jobs=-1, random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
}

reg_results, reg_pipes = [], {}
kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
for name, model in reg_models.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    cv_r2 = cross_val_score(pipe, X, y, cv=kf, scoring="r2", n_jobs=-1).mean()   # 5-fold CV
    reg_results.append({
        "model": name,
        "R2_test": r2_score(y_test, pred),
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2_cv(5-fold)": cv_r2,
    })
    reg_pipes[name] = pipe

reg_results = pd.DataFrame(reg_results).sort_values("R2_test", ascending=False)
print(reg_results.round(4).to_string(index=False))

best_reg_name = reg_results.iloc[0]["model"]
best_reg = reg_pipes[best_reg_name]
print(f"\nBest regression model: {best_reg_name}")

# 7.4 Actual vs predicted + residuals for the best model
pred = best_reg.predict(X_test)
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
ax[0].scatter(y_test, pred, alpha=0.4, s=12)
lims = [min(y_test.min(), pred.min()), max(y_test.max(), pred.max())]
ax[0].plot(lims, lims, "r--")
ax[0].set_xlabel("Actual exam_score"); ax[0].set_ylabel("Predicted exam_score")
ax[0].set_title(f"Actual vs Predicted ({best_reg_name})")
sns.histplot(y_test - pred, bins=30, kde=True, ax=ax[1])
ax[1].set_title("Residuals (actual - predicted)")
finish_plot("07_regression_actual_vs_predicted")
```

```python
# 7. REGRESSION MODELS  (predict exam_score)
# ------------------------------------------------------------------------------
header("7. REGRESSION MODELS")

# 7.1 Build feature matrix. We DROP: id + all leaky columns (derived from exam_score)
drop_cols = [c for c in [ID_COL] + LEAKY_COLS if c in df.columns]
X = df.drop(columns=drop_cols)
y = df[TARGET_REG]

num_features = X.select_dtypes(include="number").columns.tolist()
cat_features = X.select_dtypes(exclude="number").columns.tolist()
print(f"Features used: {X.shape[1]}  ({len(num_features)} numeric, {len(cat_features)} categorical)")


def make_preprocessor(scale_numeric=True):
    """Impute + scale numeric columns, impute + one-hot encode categorical columns."""
    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    if scale_numeric:
        num_steps.append(("scaler", StandardScaler()))
    return ColumnTransformer([
        ("num", Pipeline(num_steps), num_features),
        ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                          ("onehot", OneHotEncoder(handle_unknown="ignore"))]), cat_features),
    ])


# 7.2 Train / test split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)

# 7.3 Models to compare
reg_models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(n_estimators=200, n_jobs=-1, random_state=RANDOM_STATE),
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
}

reg_results, reg_pipes = [], {}
kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
for name, model in reg_models.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    cv_r2 = cross_val_score(pipe, X, y, cv=kf, scoring="r2", n_jobs=-1).mean()   # 5-fold CV
    reg_results.append({
        "model": name,
        "R2_test": r2_score(y_test, pred),
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2_cv(5-fold)": cv_r2,
    })
    reg_pipes[name] = pipe

reg_results = pd.DataFrame(reg_results).sort_values("R2_test", ascending=False)
print(reg_results.round(4).to_string(index=False))

best_reg_name = reg_results.iloc[0]["model"]
best_reg = reg_pipes[best_reg_name]
print(f"\nBest regression model: {best_reg_name}")

# 7.4 Actual vs predicted + residuals for the best model
pred = best_reg.predict(X_test)
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
ax[0].scatter(y_test, pred, alpha=0.4, s=12)
lims = [min(y_test.min(), pred.min()), max(y_test.max(), pred.max())]
ax[0].plot(lims, lims, "r--")
ax[0].set_xlabel("Actual exam_score"); ax[0].set_ylabel("Predicted exam_score")
ax[0].set_title(f"Actual vs Predicted ({best_reg_name})")
sns.histplot(y_test - pred, bins=30, kde=True, ax=ax[1])
ax[1].set_title("Residuals (actual - predicted)")
finish_plot("07_regression_actual_vs_predicted")
```

```python
# 8. CLASSIFICATION MODELS (predict pass_status / performance_level)
# ------------------------------------------------------------------------------
#header("8. CLASSIFICATION MODELS")

clf_best_pipes = {}      # keep the best pipeline for each target
clf_tables = {}

for target in TARGET_CLF:
    if target not in df.columns:
        continue
    print(f"\n{'-' * 70}\nTarget: {target}\n{'-' * 70}")

    # Features: same X as before (all leaky columns are already removed)
    le = LabelEncoder()
    y_c = le.fit_transform(df[target].astype(str))
    classes = le.classes_
    is_binary = len(classes) == 2

    Xtr, Xte, ytr, yte = train_test_split(X, y_c, test_size=0.2, stratify=y_c, random_state=RANDOM_STATE)

    clf_models = {
        "Logistic Regression": LogisticRegression(max_iter=2000),
        "Random Forest": RandomForestClassifier(n_estimators=200, n_jobs=-1, random_state=RANDOM_STATE),
        "Gradient Boosting": GradientBoostingClassifier(random_state=RANDOM_STATE),
    }

    rows, pipes = [], {}
    for name, model in clf_models.items():
        pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
        pipe.fit(Xtr, ytr)
        pred_c = pipe.predict(Xte)
        proba = pipe.predict_proba(Xte)
        avg = "binary" if is_binary else "weighted"
        auc = (roc_auc_score(yte, proba[:, 1]) if is_binary
               else roc_auc_score(yte, proba, multi_class="ovr", average="weighted"))
        rows.append({
            "model": name,
            "accuracy": accuracy_score(yte, pred_c),
            "precision": precision_score(yte, pred_c, average=avg, zero_division=0),
            "recall": recall_score(yte, pred_c, average=avg, zero_division=0),
            "f1": f1_score(yte, pred_c, average=avg, zero_division=0),
            "roc_auc": auc,
        })
        pipes[name] = pipe

    table = pd.DataFrame(rows).sort_values("f1", ascending=False)
    clf_tables[target] = table
    print(table.round(4).to_string(index=False))

    best_name = table.iloc[0]["model"]
    best_pipe = pipes[best_name]
    clf_best_pipes[target] = (best_pipe, le, Xte, yte)
    print(f"\nBest model for {target}: {best_name}")
    print("\nClassification report:")
    print(classification_report(yte, best_pipe.predict(Xte), target_names=classes, zero_division=0))

    # Confusion matrix
    fig, ax = plt.subplots(figsize=(6, 5))
    ConfusionMatrixDisplay(confusion_matrix(yte, best_pipe.predict(Xte)), display_labels=classes).plot(
        ax=ax, cmap="Blues", xticks_rotation=30)
    ax.set_title(f"Confusion matrix - {target} ({best_name})")
    finish_plot(f"08_confusion_matrix_{target}")
```

```python
# 9. FEATURE IMPORTANCE & EXPLAINABILITY
# ------------------------------------------------------------------------------
header("9. FEATURE IMPORTANCE & EXPLAINABILITY")

# 9.1 Random-forest built-in importance (fast, but biased toward numeric features)
rf_pipe = Pipeline([("prep", make_preprocessor()),
                    ("model", RandomForestRegressor(n_estimators=300, n_jobs=-1, random_state=RANDOM_STATE))])
rf_pipe.fit(X_train, y_train)
feat_names = rf_pipe.named_steps["prep"].get_feature_names_out()
imp = (pd.Series(rf_pipe.named_steps["model"].feature_importances_, index=feat_names)
       .sort_values(ascending=False))
print("Top 15 features (Random Forest importance):")
print(imp.head(15).round(4))

plt.figure(figsize=(8, 7))
imp.head(20).sort_values().plot(kind="barh")
plt.title("Top 20 feature importances (Random Forest)")
finish_plot("09_rf_feature_importance")

# 9.2 Permutation importance (more reliable; works on the ORIGINAL columns)
#perm = permutation_importance(rf_pipe, X_test, y_test, n_repeats=3,
#                              random_state=RANDOM_STATE, n_jobs=-1, scoring="r2")
perm = permutation_importance(
    rf_pipe,
    X_test,
    y_test,
    n_repeats=3,
    random_state=RANDOM_STATE,
    n_jobs=1,
    scoring="r2"
)
perm_df = (pd.DataFrame({"feature": X_test.columns,
                         "importance": perm.importances_mean,
                         "std": perm.importances_std})
           .sort_values("importance", ascending=False))
print("\nPermutation importance (drop in R2 when the feature is shuffled):")
print(perm_df.head(15).round(4).to_string(index=False))

plt.figure(figsize=(8, 7))
top = perm_df.head(20).iloc[::-1]
plt.barh(top["feature"], top["importance"], xerr=top["std"])
plt.title("Top 20 permutation importances")
finish_plot("09_permutation_importance")

# 9.3 SHAP values (optional - needs `pip install shap`)
try:
    import shap
    sample_X = X_test.sample(min(500, len(X_test)), random_state=RANDOM_STATE)
    transformed = rf_pipe.named_steps["prep"].transform(sample_X)
    if hasattr(transformed, "toarray"):
        transformed = transformed.toarray()
    explainer = shap.TreeExplainer(rf_pipe.named_steps["model"])
    shap_values = explainer.shap_values(transformed)
    shap.summary_plot(shap_values, transformed, feature_names=feat_names, show=False, max_display=20)
    finish_plot("09_shap_summary")
    print("SHAP summary plot created.")
except Exception as e:
    print("SHAP skipped:", e)
```

```python
# 10. CLUSTERING (student profiles)
# ------------------------------------------------------------------------------
header("10. CLUSTERING - STUDENT PROFILES")

cluster_features = [c for c in num_features]            # numeric features only (no leaky columns)
Xc = StandardScaler().fit_transform(SimpleImputer(strategy="median").fit_transform(df[cluster_features]))

# 10.1 Choose number of clusters: elbow (inertia) + silhouette score
ks = range(2, 9)
inertias, sils = [], []
sil_idx = np.random.RandomState(RANDOM_STATE).choice(len(Xc), min(len(Xc), 5000), replace=False)
for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=RANDOM_STATE).fit(Xc)
    inertias.append(km.inertia_)
    sils.append(silhouette_score(Xc[sil_idx], km.labels_[sil_idx]))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(list(ks), inertias, "o-"); ax[0].set_title("Elbow method"); ax[0].set_xlabel("k")
ax[1].plot(list(ks), sils, "o-", color="darkorange"); ax[1].set_title("Silhouette score"); ax[1].set_xlabel("k")
finish_plot("10_choose_k")

best_k = list(ks)[int(np.argmax(sils))]
print(f"Best k by silhouette score: {best_k}  (change N_CLUSTERS below to override)")
N_CLUSTERS = best_k

# 10.2 Fit final KMeans
km = KMeans(n_clusters=N_CLUSTERS, n_init=10, random_state=RANDOM_STATE).fit(Xc)
df["cluster"] = km.labels_

# 10.3 Profile each cluster (mean of important columns)
profile_cols = [c for c in [TARGET_REG, "study_hours_per_day", "attendance_percentage",
                            "assignment_completion_rate", "sleep_hours", "previous_exam_score"]
                if c in df.columns]
print("\nCluster sizes:")
print(df["cluster"].value_counts().sort_index())
print("\nCluster profiles (means):")
print(df.groupby("cluster")[profile_cols].mean().round(2))

# 10.4 Visualise clusters in 2D with PCA
pcs = PCA(n_components=2, random_state=RANDOM_STATE).fit_transform(Xc)
plt.figure(figsize=(8, 6))
sns.scatterplot(x=pcs[:, 0], y=pcs[:, 1], hue=df["cluster"], palette="tab10", s=18, alpha=0.7)
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.title("Student clusters (PCA projection)")
finish_plot("10_clusters_pca")

# Pass rate per cluster
if "pass_status" in df.columns:
    print("\nPass-status distribution inside each cluster:")
    print(pd.crosstab(df["cluster"], df["pass_status"], normalize="index").round(3))
```

```python
# 11. FAIRNESS / EQUITY ANALYSIS (OLS regression with statsmodels)
# ------------------------------------------------------------------------------
header("11. FAIRNESS / EQUITY ANALYSIS")
# Do gender, family income, school type or urban/rural still matter AFTER we control
# for effort (study hours, attendance, ...)?  Look at the coefficients and p-values.

demographic = [c for c in ["gender", "family_income", "urban_rural", "school_type",
                           "parent_education"] if c in df.columns]
effort = [c for c in ["study_hours_per_day", "attendance_percentage",
                      "assignment_completion_rate", "self_study_hours", "sleep_hours",
                      "previous_exam_score"] if c in df.columns]


def term(col):
    """Wrap non-numeric columns in C() so statsmodels treats them as categories."""
    return col if pd.api.types.is_numeric_dtype(df[col]) else f"C({col})"


if demographic:
    formula = f"{TARGET_REG} ~ " + " + ".join(term(c) for c in demographic + effort)
    ols = smf.ols(formula, data=df).fit()
    print(ols.summary())

    coef = (pd.DataFrame({"coef": ols.params, "p_value": ols.pvalues})
            .drop(index="Intercept"))
    sig = coef[coef["p_value"] < 0.05].sort_values("coef")
    print("\nSignificant predictors (p < 0.05):")
    print(sig.round(4) if len(sig) else "None")

    # Average score gap between demographic groups (raw, not controlled)
    for c in demographic:
        means = df.groupby(c)[TARGET_REG].mean()
        print(f"Raw score gap across '{c}': {means.max() - means.min():.2f} points")
```
